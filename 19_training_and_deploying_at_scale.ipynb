## Сохранение и загрузка `SavedModel`

````python
(X_train_full, y_train_full), (X_test, y_test) = keras.datasets.mnist.load_data()
X_train_full = X_train_full[..., np.newaxis].astype(np.float32) / 255.
X_test = X_test[..., np.newaxis].astype(np.float32) / 255.
X_valid, X_train = X_train_full[:5000], X_train_full[5000:]
y_valid, y_train = y_train_full[:5000], y_train_full[5000:]
X_new = X_test[:3]
````

````python
np.random.seed(42)
tf.random.set_seed(42)

model = keras.models.Sequential([
    keras.layers.Flatten(input_shape=[28, 28, 1]),
    keras.layers.Dense(100, activation="relu"),
    keras.layers.Dense(10, activation="softmax")
])
model.compile(loss="sparse_categorical_crossentropy",
              optimizer=keras.optimizers.SGD(learning_rate=1e-2),
              metrics=["accuracy"])
model.fit(X_train, y_train, epochs=10, validation_data=(X_valid, y_valid))
````

# Распределённое обучение (Distributed Training) <a id="Distributed-Training"></a>

````python
keras.backend.clear_session()
tf.random.set_seed(42)
np.random.seed(42)
````

````python
def create_model():
    return keras.models.Sequential([
        keras.layers.Conv2D(filters=64, kernel_size=7, activation="relu",
                            padding="same", input_shape=[28, 28, 1]),
        keras.layers.MaxPooling2D(pool_size=2),
        keras.layers.Conv2D(filters=128, kernel_size=3, activation="relu",
                            padding="same"), 
        keras.layers.Conv2D(filters=128, kernel_size=3, activation="relu",
                            padding="same"),
        keras.layers.MaxPooling2D(pool_size=2),
        keras.layers.Flatten(),
        keras.layers.Dense(units=64, activation='relu'),
        keras.layers.Dropout(0.5),
        keras.layers.Dense(units=10, activation='softmax'),
    ])
````

````python
batch_size = 100
model = create_model()
model.compile(loss="sparse_categorical_crossentropy",
              optimizer=keras.optimizers.SGD(learning_rate=1e-2),
              metrics=["accuracy"])
model.fit(X_train, y_train, epochs=10,
          validation_data=(X_valid, y_valid), batch_size=batch_size)
````

````python
keras.backend.clear_session()
tf.random.set_seed(42)
np.random.seed(42)

distribution = tf.distribute.MirroredStrategy()

# Изменить алгоритм all-reduce по умолчанию:
#distribution = tf.distribute.MirroredStrategy(
#    cross_device_ops=tf.distribute.HierarchicalCopyAllReduce())

# Задать список используемых GPU:
#distribution = tf.distribute.MirroredStrategy(devices=["/gpu:0", "/gpu:1"])

# Использовать вместо этого стратегию центрального хранения:
#distribution = tf.distribute.experimental.CentralStorageStrategy()

#if IS_COLAB and "COLAB_TPU_ADDR" in os.environ:
#  tpu_address = "grpc://" + os.environ["COLAB_TPU_ADDR"]
#else:
#  tpu_address = ""
#resolver = tf.distribute.cluster_resolver.TPUClusterResolver(tpu_address)
#tf.config.experimental_connect_to_cluster(resolver)
#tf.tpu.experimental.initialize_tpu_system(resolver)
#distribution = tf.distribute.experimental.TPUStrategy(resolver)

with distribution.scope():
    model = create_model()
    model.compile(loss="sparse_categorical_crossentropy",
                  optimizer=keras.optimizers.SGD(learning_rate=1e-2),
                  metrics=["accuracy"])
````

````python
batch_size = 100 # должно делиться на число рабочих узлов
model.fit(X_train, y_train, epochs=10,
          validation_data=(X_valid, y_valid), batch_size=batch_size)
````

````python
model.predict(X_new)
````

Собственный цикл обучения:

````python
keras.backend.clear_session()
tf.random.set_seed(42)
np.random.seed(42)

K = keras.backend

distribution = tf.distribute.MirroredStrategy()

with distribution.scope():
    model = create_model()
    optimizer = keras.optimizers.SGD()

with distribution.scope():
    dataset = tf.data.Dataset.from_tensor_slices((X_train, y_train)).repeat().batch(batch_size)
    input_iterator = distribution.make_dataset_iterator(dataset)
    
@tf.function
def train_step():
    def step_fn(inputs):
        X, y = inputs
        with tf.GradientTape() as tape:
            Y_proba = model(X)
            loss = K.sum(keras.losses.sparse_categorical_crossentropy(y, Y_proba)) / batch_size

        grads = tape.gradient(loss, model.trainable_variables)
        optimizer.apply_gradients(zip(grads, model.trainable_variables))
        return loss

    per_replica_losses = distribution.experimental_run(step_fn, input_iterator)
    mean_loss = distribution.reduce(tf.distribute.ReduceOp.SUM,
                                    per_replica_losses, axis=None)
    return mean_loss

n_epochs = 10
with distribution.scope():
    input_iterator.initialize()
    for epoch in range(n_epochs):
        print("Epoch {}/{}".format(epoch + 1, n_epochs))
        for iteration in range(len(X_train) // batch_size):
            print("\rLoss: {:.3f}".format(train_step().numpy()), end="")
        print()
````

## Обучение на нескольких серверах

Кластер TensorFlow — это группа процессов TensorFlow, работающих параллельно (обычно на разных машинах) и взаимодействующих друг с другом для выполнения некоторой работы, например обучения или выполнения нейронной сети. Каждый процесс TF в кластере называется «задачей» (task) или «сервером TF» (TF server). У него есть IP-адрес, порт и тип (также называемый ролью или заданием, job). Тип может быть `"worker"`, `"chief"`, `"ps"` (сервер параметров) или `"evaluator"`:
* Каждый **рабочий узел** (worker) выполняет вычисления, обычно на машине с одним или несколькими GPU.
* **Главный узел** (chief) тоже выполняет вычисления, но, кроме того, берёт на себя дополнительную работу, например запись журналов TensorBoard или сохранение контрольных точек. В кластере только один главный узел, обычно это первый рабочий узел (то есть worker #0).
* **Сервер параметров** (parameter server, ps) только хранит значения переменных; обычно он находится на машине только с CPU.
* **Оценщик** (evaluator), как понятно из названия, занимается оценкой. Обычно в кластере один оценщик.

Множество задач одного типа часто называют «заданием» (job). Например, задание «worker» — это множество всех рабочих узлов.

Чтобы запустить кластер TensorFlow, его сначала нужно определить. Это означает указать все задачи (IP-адрес, TCP-порт и тип). Например, следующая спецификация кластера определяет кластер из 3 задач (2 рабочих узла и 1 сервер параметров). Это словарь с одним ключом на каждое задание, а значения — списки адресов задач:

````python
cluster_spec = {
    "worker": [
        "machine-a.example.com:2222",  # /job:worker/task:0
        "machine-b.example.com:2222"   # /job:worker/task:1
    ],
    "ps": ["machine-c.example.com:2222"] # /job:ps/task:0
}
````

Каждая задача в кластере может взаимодействовать с любой другой задачей, поэтому настройте межсетевой экран так, чтобы разрешить любой обмен данными между этими машинами на этих портах (обычно проще использовать один и тот же порт на всех машинах).

При запуске задаче нужно сообщить, какая она: её тип и индекс (индекс задачи также называют идентификатором задачи). Распространённый способ указать всё сразу (и спецификацию кластера, и тип и идентификатор текущей задачи) — задать переменную окружения `TF_CONFIG` перед запуском программы. Это должен быть закодированный в JSON словарь, содержащий спецификацию кластера (под ключом `"cluster"`), а также тип и индекс запускаемой задачи (под ключом `"task"`). Например, следующая переменная окружения `TF_CONFIG` определяет тот же кластер, что и выше, с 2 рабочими узлами и 1 сервером параметров, и указывает, что запускаемая задача — это worker #1:

````python
import os
import json

os.environ["TF_CONFIG"] = json.dumps({
    "cluster": cluster_spec,
    "task": {"type": "worker", "index": 1}
})
os.environ["TF_CONFIG"]
````

Теперь запустим более простой кластер всего из двух рабочих задач, обе на локальной машине. Для обучения модели на этих двух задачах используем `MultiWorkerMirroredStrategy`.

Первый шаг — написать код обучения. Поскольку этот код будет использоваться для запуска обоих рабочих узлов, каждого в своём процессе, мы записываем его в отдельный файл Python `my_mnist_multiworker_task.py`. Код относительно простой, но стоит обратить внимание на пару важных моментов:
* Мы создаём `MultiWorkerMirroredStrategy` прежде, чем делать с TensorFlow что-либо ещё.
* Журналированием в TensorBoard и сохранением контрольных точек будет заниматься только один из рабочих узлов. Как уже упоминалось, этот узел называется *главным* (chief), и по соглашению обычно это worker #0.

````python
%%writefile my_mnist_multiworker_task.py

import os
import numpy as np
import tensorflow as tf
from tensorflow import keras
import time

# В начале программы
distribution = tf.distribute.MultiWorkerMirroredStrategy()

resolver = tf.distribute.cluster_resolver.TFConfigClusterResolver()
print("Starting task {}{}".format(resolver.task_type, resolver.task_id))

# Только worker #0 будет записывать контрольные точки и журналы TensorBoard
if resolver.task_id == 0:
    root_logdir = os.path.join(os.curdir, "my_mnist_multiworker_logs")
    run_id = time.strftime("run_%Y_%m_%d-%H_%M_%S")
    run_dir = os.path.join(root_logdir, run_id)
    callbacks = [
        keras.callbacks.TensorBoard(run_dir),
        keras.callbacks.ModelCheckpoint("my_mnist_multiworker_model.h5",
                                        save_best_only=True),
    ]
else:
    callbacks = []

# Загружаем и готовим набор данных MNIST
(X_train_full, y_train_full), (X_test, y_test) = keras.datasets.mnist.load_data()
X_train_full = X_train_full[..., np.newaxis] / 255.
X_valid, X_train = X_train_full[:5000], X_train_full[5000:]
y_valid, y_train = y_train_full[:5000], y_train_full[5000:]

with distribution.scope():
    model = keras.models.Sequential([
        keras.layers.Conv2D(filters=64, kernel_size=7, activation="relu",
                            padding="same", input_shape=[28, 28, 1]),
        keras.layers.MaxPooling2D(pool_size=2),
        keras.layers.Conv2D(filters=128, kernel_size=3, activation="relu",
                            padding="same"), 
        keras.layers.Conv2D(filters=128, kernel_size=3, activation="relu",
                            padding="same"),
        keras.layers.MaxPooling2D(pool_size=2),
        keras.layers.Flatten(),
        keras.layers.Dense(units=64, activation='relu'),
        keras.layers.Dropout(0.5),
        keras.layers.Dense(units=10, activation='softmax'),
    ])
    model.compile(loss="sparse_categorical_crossentropy",
                  optimizer=keras.optimizers.SGD(learning_rate=1e-2),
                  metrics=["accuracy"])

model.fit(X_train, y_train, validation_data=(X_valid, y_valid),
          epochs=10, callbacks=callbacks)
````

В реальном приложении обычно на каждой машине работает один рабочий узел, но в этом примере оба рабочих узла запущены на одной машине, поэтому оба будут пытаться использовать всю доступную память GPU (если на машине есть GPU), и это, скорее всего, приведёт к ошибке нехватки памяти (OOM). Чтобы этого избежать, можно с помощью переменной окружения `CUDA_VISIBLE_DEVICES` назначить каждому рабочему узлу свой GPU. Или можно просто отключить поддержку GPU, вот так:

````python
os.environ["CUDA_VISIBLE_DEVICES"] = "-1"
````
