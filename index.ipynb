## Предварительные требования <a id="Prerequisites"></a>

### Чтобы понимать материал
* **Python** – не нужно быть экспертом в программировании на Python, но основы знать необходимо. Если вы их не знаете, хорошая отправная точка — официальное [руководство по Python](https://docs.python.org/3/tutorial/).
* **Научный Python** – мы будем использовать несколько популярных библиотек Python, в частности NumPy, matplotlib и pandas. Если вы с ними не знакомы, вероятно, стоит начать с руководств из раздела инструментов (особенно по NumPy).
* **Математика** – мы также будем использовать некоторые понятия линейной алгебры, математического анализа, статистики и теории вероятностей. Если вы когда-то это изучали, то сможете следить за изложением, поскольку оно не будет очень сложным; но если вы не знакомы с этими темами или вам нужно освежить знания, пройдите соответствующее введение в разделе математики.

### Чтобы запускать примеры
* **Jupyter** – эти ноутбуки основаны на Jupyter. Их можно запустить одним щелчком на облачной платформе, такой как Binder, Deepnote или Colaboratory (установка не требуется), можно просто просматривать их в программе просмотра Jupyter.org, а можно установить всё на свою машину — как вам удобнее. Подробнее см. на [главной странице](https://github.com/ageron/handson-ml2/).
